```python
import torch
x = torch.randn(1, 3, 6)
x
```

```python
x = x.softmax(dim=1)
x_sorted, indices = torch.sort(x, dim=1, descending=True)
x
```

```python
x_sorted
```

```python
indices
```

```python
weights = torch.logspace(start=0, end=3-1, steps=3, base=0.9).unsqueeze(-1)
weighted_sorted = x_sorted * weights.to(x.device)
```

```python
weighted_sorted
```

```python
result = torch.empty_like(x)
inv_indices = indices.argsort(dim=1)
result.scatter_(1, inv_indices, weighted_sorted)
```
