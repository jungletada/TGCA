```python
import os
import shutil
import random

# Paths to your folders and list file
dir_segaug = "datasets/VOCdevkit/VOC2012/SegmentationClassAug"
dir_pseudo = "results_voc/mctgformer_448/pseudo_mask"
dir_combine = "results_voc/mctgformer_448/model_mask"
list_id_path = "configs/voc12/train_aug_id.txt"

# Ensure the Combine directory exists
if not os.path.exists(dir_combine):
    os.makedirs(dir_combine)

# Read file names from list_id.txt
with open(list_id_path, 'r') as f:
    file_names = f.read().splitlines()

# Shuffle and split file names based on the specified ratio
random.shuffle(file_names)
split_index = int(len(file_names) * 0.3)
segaug_files = file_names[:split_index]
pseudo_files = file_names[split_index:]

# Function to copy files from source to destination
def copy_files(file_list, source_dir, dest_dir):
    for file_name in file_list:
        src_file_path = os.path.join(source_dir, file_name + '.png')
        dest_file_path = os.path.join(dest_dir, file_name + '.png')
        if os.path.exists(src_file_path):
            shutil.copy(src_file_path, dest_file_path)
        else:
            raise ValueError(f"{src_file_path} not exist!")

# Copy selected files to the Combine folder
copy_files(segaug_files, dir_segaug, dir_combine)
copy_files(pseudo_files, dir_pseudo, dir_combine)

print("Files have been successfully copied to the Combine folder.")
```
