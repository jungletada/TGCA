```python
import torch

cls_tokens = torch.randn(1, 1, 384)
ext_tokens = torch.randn(4, 20, 384)
```

```python
ext = torch.mean(ext_tokens, dim=1, keepdim=True)
combine = ext + cls_tokens
combine.shape
```

```python
import torch.nn as nn
extend_cls = nn.Conv1d(1, 20, kernel_size=1, stride=1)
combine_ext = extend_cls(combine)
combine_ext.shape
```
